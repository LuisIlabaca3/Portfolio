```python
import google.generativeai as genai
from diffusers import StableDiffusionPipeline
import torch
from flask import Flask, request, render_template_string
from PIL import Image
import io
import base64
from pyngrok import ngrok

# Configure Gemini API
API_KEY = "AIzaSyAtOVRZbGSEyXBE64jkjD3_qXTfvfuwdOc" # Replace with your key
genai.configure(api_key=API_KEY)
model = genai.GenerativeModel("gemini-2.0-pro-exp")

# Load Stable Diffusion
pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)
pipe = pipe.to("cuda")

# Initialize Flask app
app = Flask(__name__)

# HTML template
HTML_TEMPLATE = """
<!DOCTYPE html>
<html>
<head><title>Pitbull Info App</title></head>
<body>
    <h1>Get Info and Image</h1>
    <form method="POST">
        <label>Enter your prompt:</label><br>
        <textarea name="prompt" rows="4" cols="50">Tell me about American Rednose Pitbulls.</textarea><br>
        <input type="submit" value="Generate">
    </form>
    {% if text %}
        <h2>Generated Text:</h2>
        <p>{{ text }}</p>
    {% endif %}
    {% if image %}
        <h2>Generated Image:</h2>
        <img src="data:image/png;base64,{{ image }}" alt="Generated Image">
    {% endif %}
</body>
</html>
"""

@app.route("/", methods=["GET", "POST"])
def home():
    text = None
    image_base64 = None

    if request.method == "POST":
        prompt = request.form["prompt"]

        # Generate text with Gemini
        response = model.generate_content(
            prompt,
            generation_config=genai.types.GenerationConfig(
                max_output_tokens=200,
                temperature=0.7,
                top_p=0.9
            )
        )
        text = response.text

        # Create image prompt
        image_prompt = f"{prompt.split('.')[0]}, detailed, realistic"

        # Generate image
        image = pipe(image_prompt, num_inference_steps=30, guidance_scale=7.5).images[0]

        # Convert to base64
        buffered = io.BytesIO()
        image.save(buffered, format="PNG")
        image_base64 = base64.b64encode(buffered.getvalue()).decode("utf-8")

    return render_template_string(HTML_TEMPLATE, text=text, image=image_base64)

# Set up ngrok
ngrok.set_auth_token("2tMWD5jRTtynzA9kRKC6n67fV1c_4FdiLxE9FD9XQHky1iSS8")
public_url = ngrok.connect(5000)
print(f"Public URL: {public_url}")

# Run Flask app
if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
```
